```python
import pandas as pd

orders = pd.read_csv("../sample_data/orders.csv")
orders.head()        # 先頭5行を表示して列と値の雰囲気を掴む
```

```python
orders.info()        # 列の型と欠損の有無を確認する
```

```python
orders.describe()    # 数値列の分布（平均・最小・最大など）を確認する
```

```python
# ステータス別の注文数
orders["status"].value_counts()
```

```python
# 月別の注文数
orders["created_at"] = pd.to_datetime(orders["created_at"])
orders.groupby(orders["created_at"].dt.to_period("M")).size()
```

```python
import plotly.express as px

monthly = orders.groupby(orders["created_at"].dt.to_period("M")).size().reset_index(name="count")
monthly["created_at"] = monthly["created_at"].astype(str)
px.bar(monthly, x="created_at", y="count")
```
